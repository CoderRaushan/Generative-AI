```python
graph=build_blog_graph()

config={"configurable":{"thread_id":"1"}}

state=graph.invoke({
    "topic":"What is MCP?",
    "audience":"Ai Developer",
},
config=config
)

snap=graph.get_state(config)
interrupt_payload=snap.interrupts[0].value
print("Stage: ", interrupt_payload.get("stage"))
print("\n -------Research Outline for your Review--------")
print(interrupt_payload.get("research"))
```

```python
state=graph.invoke(
    Command(resume={"action":"approve","feedback":""}),
    config=config
)   
state
```
