#### Text Document Loader

```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/TextData.txt")

textData=loader.load()
print(textData[0].page_content)
```

#### Pdf Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader=PyPDFLoader("../data/PdfData.pdf")

pdfData=loader.load()
print(pdfData[1].page_content)
```

#### CSV Document loader 

```python
from langchain_community.document_loaders import CSVLoader
loader=CSVLoader("../data/CsvData.csv")

csvData=loader.load()
csvData
```

#### Web Document Loader

```python
from langchain_community.document_loaders import WebBaseLoader
loader=WebBaseLoader(web_path="https://rausnotes39.netlify.app/")

webData=loader.load()
print(webData[0].page_content)
```

#### Wikipedia loader 

```python
from langchain_community.document_loaders import WikipediaLoader

loader = WikipediaLoader(query="Gen AI",load_max_docs=2)

wikiData = loader.load()
print(wikiData)
```
