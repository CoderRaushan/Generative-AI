```python
from langchain_openai import ChatOpenAI
from IPython.display import display, Markdown

llm = ChatOpenAI(model="gpt-4o")

prompts=[
    {
        "role":"system",
        "content":"your are my translator, and translate the input into hinglish"
    },
    {
        "role":"user",
        "content":"I love playing cricket, and my favourite player is MS Dhoni and i am big fan of csk ipl team."
    }
]

res = llm.invoke(prompts)

display(Markdown(res.content))
```

```python
from langchain_openai import ChatOpenAI
from IPython.display import display, Markdown
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model="gpt-4o")

prompts=ChatPromptTemplate.from_messages([
    {
        "role":"system",
        "content":"your are my translator, and translate the input into {language}"
    },
    {
        "role":"user",
        "content":"{query}"
    }
])
final_prompts=prompts.format_messages(language="hindi", query="I love playing cricket, and my favourite player is MS Dhoni and i am big fan of csk ipl team.")

res = llm.invoke(final_prompts)

display(Markdown(res.content))
```

```python
from langchain_openai import ChatOpenAI
from IPython.display import display, Markdown
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOpenAI(model="gpt-4o")
output = StrOutputParser()

prompts=ChatPromptTemplate.from_messages([
    {
        "role":"system",
        "content":"your are my translator, and translate the input into {language}"
    },
    {
        "role":"user",
        "content":"{query}"
    }
])

chains = prompts | llm | output
res=chains.invoke({"language":"Hindi","query":"I love you"})
display(Markdown(res))
```
