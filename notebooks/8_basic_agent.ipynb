```python
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent


@tool
def add_numbers(a:int,b:int):
    """
        It will return sum of two numbers.

        Args:
            a:Number One
            b:Number Two
    """
    return a+b



@tool
def multiply_two_numbers(a:int,b:int):
    """
        It will return multiply of two numbers.

        Args:
            a:Number One
            b:Number Two
    """
    return a*b




llm=ChatOpenAI(model="gpt-4")

Agent=create_agent(
    model=llm,
    tools=[add_numbers,multiply_two_numbers],
    system_prompt="You are my maths teacher, and always use tool for calculation."
)
res=Agent.invoke({"messages":[{"role":"user","content":"what is 2+5*6 ?"}]})
print(res["messages"][-1].content)


```
